```python
import pandas as pd
import numpy as np

# One way of running regressions
import statsmodels.formula.api as smf
```

```python
# let's you use random function and replicate
np.random.seed(42)

n = 10
p = 3

# generate random n draw from a normal distribution N(0, I_p) (p is dimensionality)

np.random.normal(size=(n, p))
```

Suppose $X \sim N(0, \mathbb{I}_p)$ and $Y \sim N(0, 1)$ are statistically independent. It follows that the best linear predictor of $Y$ is $\beta^\prime X$. $R^2_{\text{pop}}$, the proportion of variation of $Y$ explained by the BLP, is 0 ($R^2_{\text{pop}} = 0$).

```python
# let's you use random function and replicate
np.random.seed(42)

Y = np.random.normal(size = (n, 1))
X = np.random.normal(size = (n, p))
```

**Question:** What does the $R^2_{\text{sample}}$ look like for different values of $n$ and $p$? First think of the components we need to estimate to determine that

```python
data_dict = {'Y': Y, 'X': X}

# estimate using OLS
ols_mod = smf.ols('Y ~ X', data={'Y': Y, 'X': X})
ols_fit = ols_mod.fit()

# find the R^2
r_squared = ols_fit.rsquared
r_squared_adj = ols_fit.rsquared_adj

# print these values, so you can see them
print(r_squared)
# print(r_squared_adj)
```

Now put all of these steps into a function, so we can easily repeat this for different values of $n$ and $p$. Notice that I 

```python
def regression_stats(n, p):
    np.random.seed(123)
    X = np.random.normal(size=(n, p))
    y = np.random.normal(size=(n, 1))

    data_dict = {'y': y, 'X': X}

    # estimate using OLS
    ols_mod = smf.ols('y ~ X', data={'y': y, 'X': X})
    ols_fit = ols_mod.fit()

    p_div_n = p / n
    mse = ols_fit.mse_resid
    r_squared = ols_fit.rsquared
    r_squared_adj = ols_fit.rsquared_adj

    print(f'p = {p:,.0f}; n = {n:,.0f}')
    print(f"p/n = {p/n}")
    print(f"R^2 in the sample = {r_squared:.04f}")
    print(f'R^2 adjuted in the sample = {r_squared_adj:.04f}')
```

```python
regression_stats(1000, 1)
```

```python
regression_stats(n=1000, p=50)
```

```python
regression_stats(n=1000, p=500)
```
